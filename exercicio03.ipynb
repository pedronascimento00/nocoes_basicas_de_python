```python
def mario_piramid(altura):
    #verificar se a altura é valida 
    if altura < 1 or altura > 8:
        print("A altura de estar entre 1 e 8")
    #Gerar piramide 
    for i in range(1 , altura + 1):
        espaços = " " * (altura - i)
        bloco = '#' * i
        print(bloco + espaços)
altura = 5
mario_piramid(altura)
```
